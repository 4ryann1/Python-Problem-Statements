# Mathematical Opertions on Series

```python
s = pd.Series([1,2,3,4,5])
print("To Multiply all values in a series")
print("-----------------------------------")
print(s*2)
print("To find the Square of all values in a series")
print("----------------------------------------------")
print(s**2)
print("To print all values in a series that are greater than 2")
print(s[s>2])
```

# Mathematical Operation - 2

```python
series1 = pd.Series([1,2,3,4,5], index = ['A', 'B', 'C', 'D', 'E'])
series2 = pd.Series([10,20,30,40,50], index = ['A', 'B', 'C', 'D', 'E'])
series3 = pd.Series([100,200,300,400], index = ['A', 'B', 'C', 'D'])
print("To Add Series 1 and Series 2")
print("--------------------------------")
print(series1 + series2)
print("To Add Series 2 and Series 3")
print("--------------------------------")
print(series2 + series3)
print("To add series2 & series3 and filled non matching index with 0")
print(series2.add(series3,fill_value=0))
```

# Head and Tail Functions in Series

```python
#head(): It is used to access the first 5 rows of a series
arr = np.array([10,15,18,22,55,77,42,48,97])
#Create a series from array
series = pd.Series(arr)
# To print first 5 rows
print(series.head(5))
# To print first 3 rows
print(series.head(5))
```

# Tail Function - tail()

```python
arr = np.array([10,15,18,22,55,77,42,48,97])
# Create a series from array
series4 = pd.Series(arr)
# Printing last 5 rows
print(series4.tail(5)) # If no arguments are passed then, last 5 rows will be printed
# Printing last 3 rows
print(series4.tail(3))
```
